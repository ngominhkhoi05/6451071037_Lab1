```python
# Ngô Minh Khôi - 6451071037
# Bài 2 - Câu 5: Thống kê thời lượng cuộc gọi (duration) và số lần liên lạc (campaign)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Ngô Minh Khôi - 6451071037")
df = pd.read_csv(r"dataset/bank/bank.csv", sep=';')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df, x='y', y='duration', ax=axes[0], palette='pastel')
axes[0].set_title('Thời lượng cuộc gọi (giây) theo quyết định (y)')

sns.boxplot(data=df, x='y', y='campaign', ax=axes[1], palette='pastel')
axes[1].set_ylim(0, 15)
axes[1].set_title('Số lần liên lạc (campaign) theo quyết định (y)')

plt.tight_layout()
plt.savefig(r"figures/p1_b2_t28_c5.png", dpi=150)
plt.show()

print("""Nhận xét:
1. Thời lượng cuộc gọi (duration) là yếu tố quyết định hàng đầu: Khách hàng đồng ý đăng ký (y='yes') có thời gian nói chuyện trung bình dài hơn rất nhiều so với nhóm từ chối.
2. Số lần liên lạc (campaign): Tỷ lệ thành công cao nhất tập trung ở 1 đến 3 cuộc gọi đầu tiên. Việc gọi quá nhiều lần (>5 lần) không mang lại hiệu quả cao.""")
```
