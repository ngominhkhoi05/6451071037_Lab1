```python
# Ngô Minh Khôi - 6451071037
# Bài 6 - Câu 5: Lập bảng so sánh số lượng tập phổ biến và thời gian xử lý
import pandas as pd
import time
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth

print("Ngô Minh Khôi - 6451071037")
transactions = []
with open(r"dataset/groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [i.strip() for i in line.strip().split(',') if i.strip()]
        if items:
            transactions.append(items)

te = TransactionEncoder()
df_groceries = pd.DataFrame(te.fit(transactions).transform(transactions), columns=te.columns_)

supports = [0.01, 0.008, 0.006, 0.004, 0.002]
data_compare = []

for sup in supports:
    t0 = time.time()
    res_ap = apriori(df_groceries, min_support=sup, use_colnames=True)
    t_ap = time.time() - t0
    
    t0 = time.time()
    res_fp = fpgrowth(df_groceries, min_support=sup, use_colnames=True)
    t_fp = time.time() - t0
    
    data_compare.append({
        'min_support (%)': f"{sup*100:.2f}%",
        'Tập phổ biến (Apriori)': len(res_ap),
        'Thời gian Apriori (s)': round(t_ap, 4),
        'Tập phổ biến (FP-Growth)': len(res_fp),
        'Thời gian FP-Growth (s)': round(t_fp, 4),
        'Tỷ lệ tăng tốc (lần)': round(t_ap / t_fp, 2) if t_fp > 0 else 0
    })

df_compare = pd.DataFrame(data_compare)
print("--- BẢNG SO SÁNH THỜI GIAN VÀ SỐ LƯỢNG TẬP PHỔ BIẾN ---")
print(df_compare.to_string(index=False))
```
