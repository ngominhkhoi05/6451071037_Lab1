```python
# Ngô Minh Khôi - 6451071037
# Bài 2 - Câu 6: Tiền xử lý dữ liệu Bank Marketing
import pandas as pd
import numpy as np

print("Ngô Minh Khôi - 6451071037")
df = pd.read_csv(r"dataset/bank/bank.csv", sep=';')

df['y'] = df['y'].map({'no': 0, 'yes': 1}).astype(int)

binary_cols = ['default', 'housing', 'loan']
for col in binary_cols:
    df[col] = df[col].map({'no': 0, 'yes': 1}).astype(int)

df['age_group'] = pd.cut(df['age'], bins=[0, 30, 40, 50, 60, 100], labels=[0, 1, 2, 3, 4]).astype(int)
df['duration_group'] = pd.qcut(df['duration'], q=5, labels=[0, 1, 2, 3, 4]).astype(int)

val_min = df['balance'].min()
val_max = df['balance'].max()
df['balance_norm'] = (df['balance'] - val_min) / (val_max - val_min)

categorical_cols = ['job', 'marital', 'education', 'contact', 'poutcome']
df_processed = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

print("--- 5 Dòng đầu của tập dữ liệu sau tiền xử lý ---")
print(df_processed.head())
print("\nKích thước dữ liệu sau khi mã hóa:", df_processed.shape)
```
