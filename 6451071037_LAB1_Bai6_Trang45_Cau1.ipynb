```python
# Ngô Minh Khôi - 6451071037
# Bài 6 - Câu 1: Đọc dữ liệu Groceries và thống kê tổng quan
import pandas as pd

print("Ngô Minh Khôi - 6451071037")
transactions = []
with open(r"dataset/groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [i.strip() for i in line.strip().split(',') if i.strip()]
        if items:
            transactions.append(items)

unique_items = set(item for tx in transactions for item in tx)

print(f"Tổng số giao dịch trong dữ liệu Groceries: {len(transactions)}")
print(f"Tổng số mặt hàng phân biệt (unique items): {len(unique_items)}")
print("--- 5 giao dịch đầu tiên ---")
for i in range(5):
    print(f"Giao dịch {i+1}: {transactions[i]}")
```
