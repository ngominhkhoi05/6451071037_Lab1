```python
# Ngô Minh Khôi - 6451071037
# Bài 2 - Câu 4: Khảo sát Nghề nghiệp (job), Học vấn (education) với Quyết định (y)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Ngô Minh Khôi - 6451071037")
df = pd.read_csv(r"dataset/bank/bank.csv", sep=';')

plt.figure(figsize=(12, 5))
job_order = df['job'].value_counts().index
sns.countplot(data=df, y='job', hue='y', order=job_order, palette='viridis')
plt.title('Tỷ lệ đăng ký theo ngành nghề (Job)')
plt.tight_layout()
plt.savefig(r"figures/p1_b2_t28_c4_job.png", dpi=150)
plt.show()

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='education', hue='y', palette='magma')
plt.title('Tỷ lệ đăng ký theo trình độ học vấn (Education)')
plt.tight_layout()
plt.savefig(r"figures/p1_b2_t28_c4_edu.png", dpi=150)
plt.show()

print("""Nhận xét:
1. Nhóm nghề quản lý (management), kỹ thuật viên (technician) và nhân viên văn phòng (admin) có lượng tiếp cận và chấp nhận gửi tiền lớn nhất. Đặc biệt, người về hưu (retired) và sinh viên (student) có tỷ lệ phản hồi 'yes' rất tích cực.
2. Khách hàng có trình độ học vấn đại học (tertiary) và trung học (secondary) có tỷ lệ chấp thuận gửi tiết kiệm cao hơn rõ rệt so với nhóm tiểu học (primary).""")
```
