```python
# Ngô Minh Khôi - 6451071037
# Bài 2 - Câu 3: Khảo sát Độ tuổi (age), Tình trạng hôn nhân (marital) và Đăng ký (y)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Ngô Minh Khôi - 6451071037")
df = pd.read_csv(r"dataset/bank/bank.csv", sep=';')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=df, x='age', hue='y', multiple='stack', bins=30, ax=axes[0], palette='Set2')
axes[0].set_title('Phân bố độ tuổi theo quyết định gửi tiền (y)')

sns.countplot(data=df, x='marital', hue='y', ax=axes[1], palette='Set2')
axes[1].set_title('Số lượng đăng ký theo tình trạng hôn nhân (marital)')

plt.tight_layout()
plt.savefig(r"figures/p1_b2_t28_c3.png", dpi=150)
plt.show()

print("""Nhận xét:
1. Về độ tuổi: Khách hàng trong độ tuổi từ 30 đến 50 tuổi chiếm đa số các cuộc liên lạc. Nhóm khách hàng trẻ tuổi (<30) và nhóm người cao tuổi (>60) có tỷ lệ chấp nhận gửi tiền (tỷ lệ 'yes') tương đối cao hơn.
2. Về tình trạng hôn nhân: Nhóm người đã kết hôn (married) là đối tượng tiếp cận nhiều nhất, nhưng nhóm người độc thân (single) lại có tỷ lệ đồng ý gửi tiền nhỉnh hơn.""")
```
