```python
# Ngô Minh Khôi - 6451071037
# Bài 2 - Câu 2: Khảo sát các giá trị thiếu hoặc 'unknown'
import pandas as pd

print("Ngô Minh Khôi - 6451071037")
df = pd.read_csv(r"dataset/bank/bank.csv", sep=';')

unknown_counts = {}
for col in df.columns:
    count = (df[col] == 'unknown').sum()
    if count > 0:
        unknown_counts[col] = [count, round(count / len(df) * 100, 2)]

df_unknown = pd.DataFrame.from_dict(unknown_counts, orient='index', columns=['Total Unknown', '% Unknown'])
df_unknown = df_unknown.sort_values(by='Total Unknown', ascending=False)

print("Danh sách các thuộc tính có giá trị chưa xác định ('unknown'):")
print(df_unknown)
```
