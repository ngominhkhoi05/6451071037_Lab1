```python
# Ngô Minh Khôi - 6451071037
# Bài 6 - Câu 3: Tìm tập phổ biến bằng giải thuật Apriori với các min_sup
import pandas as pd
import time
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

print("Ngô Minh Khôi - 6451071037")
transactions = []
with open(r"dataset/groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [i.strip() for i in line.strip().split(',') if i.strip()]
        if items:
            transactions.append(items)

te = TransactionEncoder()
df_groceries = pd.DataFrame(te.fit(transactions).transform(transactions), columns=te.columns_)

supports = [0.01, 0.008, 0.006, 0.004, 0.002]
ap_results = []

print("Đang chạy Apriori qua các ngưỡng min_support...")
for sup in supports:
    t0 = time.time()
    res = apriori(df_groceries, min_support=sup, use_colnames=True)
    elapsed = time.time() - t0
    ap_results.append({'min_sup': sup, 'count': len(res), 'time': elapsed})
    print(f"min_sup = {sup*100:.2f}%: {len(res)} tập phổ biến, thời gian = {elapsed:.4f}s")
```
