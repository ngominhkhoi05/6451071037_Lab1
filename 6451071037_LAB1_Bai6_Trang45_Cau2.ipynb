```python
# Ngô Minh Khôi - 6451071037
# Bài 6 - Câu 2: Chuyển đổi dữ liệu về ma trận One-hot encoding bằng TransactionEncoder
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder

print("Ngô Minh Khôi - 6451071037")
transactions = []
with open(r"dataset/groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [i.strip() for i in line.strip().split(',') if i.strip()]
        if items:
            transactions.append(items)

te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_groceries = pd.DataFrame(te_ary, columns=te.columns_)

print(f"Kích thước ma trận One-hot: {df_groceries.shape}")
print("Mẫu 5 dòng đầu tiên:")
print(df_groceries.iloc[:5, :8])
```
